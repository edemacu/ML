## Core takeaway

The scalar computations of many neurons can be collected into

$$A=g(XW+B).$$

This transition from neuron-by-neuron operations to matrix operations is **vectorization**, a fundamental idea behind efficient neural-network implementations in NumPy, TensorFlow, and other modern ML frameworks.
