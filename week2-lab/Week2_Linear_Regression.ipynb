# 5. Squared-Error Cost Function

A model needs a single numerical measure of how well it fits **all** training examples.

For linear regression, we use the squared-error cost:

$
J(w,b)=\frac{1}{2m}\sum_{i=1}^{m}
\left(\hat y^{(i)}-y^{(i)}\right)^2
$

where:

- \(m\) = number of training examples,
- \($\hat y^{(i)}$\) = predicted value,
- \($y^{(i)}$\) = actual value.

### Why square the errors?

Squaring:

1. prevents positive and negative errors from cancelling,
2. penalizes large errors more heavily,
3. gives a smooth objective function that is convenient to optimize.

The factor \($\frac{1}{2}$\) simplifies derivatives; it does not change the location of the minimum.

# 8. Gradients of the Cost Function

Gradient descent requires the slope of the cost function with respect to each parameter.

For:

$
\hat y^{(i)}=wx^{(i)}+b
$

the partial derivatives are:

$$
\frac{\partial J}{\partial w}
=
\frac{1}{m}
\sum_{i=1}^{m}
(\hat y^{(i)}-y^{(i)})x^{(i)}
$$

and

$$
\frac{\partial J}{\partial b} = \frac{1}{m} \sum_{i=1}^{m} (\hat{y}^{(i)} - y^{(i)})
$$

The gradient vector is:

$$
\nabla J(w,b) =
\begin{bmatrix}
\frac{\partial J}{\partial w}\\
\frac{\partial J}{\partial b}
\end{bmatrix}
$$

The gradient points in the direction of steepest increase. Therefore, gradient descent moves in the **opposite direction**.

# 9. Gradient Descent Parameter Updates

For learning rate \($\alpha$\), gradient descent updates the parameters using:

$$
w_{\text{new}}
=
w-\alpha\frac{\partial J}{\partial w}
$$

$$
b_{\text{new}}
=
b-\alpha\frac{\partial J}{\partial b}
$$

## Important: simultaneous update

Both new values must be calculated using the **same old parameter values**.

Correct conceptual sequence:

```text
old w, old b
      ↓
compute both gradients
      ↓
compute new w and new b
      ↓
replace w and b together
```

Do not update `w` first and then compute the derivative for `b` using the new `w`.

# 11. Batch Gradient Descent

**Batch Gradient Descent** computes the gradient using **all training examples before every parameter update**.

If the dataset has \(m\) observations, one BGD iteration uses all \(m\) examples.

Conceptually:

```text
Entire training set
        ↓
Predictions for all m examples
        ↓
Gradient computed using all m examples
        ↓
Update w and b
        ↓
Repeat
```

### Advantages

- smooth, deterministic updates,
- exact gradient of the empirical training objective,
- easy to reason about.

### Limitations

- each update may be expensive for very large datasets,
- requires processing the full dataset before every update,
- can be slower in wall-clock time than stochastic or mini-batch variants.

> **Important nuance:** Batch gradient descent does not, by itself, guarantee a global minimum for every machine-learning model. For ordinary linear regression with squared-error loss, however, the objective is convex, so gradient descent can reach the global optimum when configured appropriately.

## Week 2 Summary

The core learning loop developed in this practical is:

$$
\boxed{
\text{Data}
\rightarrow
\text{Model}
\rightarrow
\text{Prediction}
\rightarrow
\text{Error}
\rightarrow
\text{Cost}
\rightarrow
\text{Gradient}
\rightarrow
\text{Parameter Update}
\rightarrow
\text{Repeat}
}
$$

This simple loop is foundational. More sophisticated machine-learning and deep-learning models use the same broad optimization idea, although their models may contain millions or billions of parameters.
