## Week 4 Summary

$$
\mathbf{x}
\rightarrow
z=\mathbf{w}^T\mathbf{x}+b
\rightarrow
\sigma(z)
\rightarrow
P(y=1\mid\mathbf{x})
\rightarrow
\text{threshold}
\rightarrow
\hat y
$$

Training:

$$
\text{logistic loss}
\rightarrow
J(w,b)
\rightarrow
\nabla J
\rightarrow
\text{gradient descent}
$$

Regularization:

$$
J_{\mathrm{reg}}
=
J
+
\frac{\lambda}{2m}\sum_jw_j^2
$$

The goal is not merely to fit the training data, but to learn a classifier that **generalizes well to unseen examples**.
