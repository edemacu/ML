## 14. Conditional Statements

Conditional statements allow a program to make decisions.

Python uses:

- `if`
- `elif`

````python
score = 76

if score >= 80:
    grade = "A"
elif score >= 70:
    grade = "B"
elif score >= 60:
    grade = "C"
else:
    grade = "Below C"

print("Grade:", grade)
````

## 15. `for` Loops

A `for` loop repeats an operation for each item in a sequence.

This is useful for iterating through datasets, features, predictions, or files.

````python
scores = [65, 72, 81, 90]

for score in scores:
    print(f"Score = {score}, squared = {score ** 2}")
````

## 16. `range()`

`range()` generates integer sequences and is commonly used with loops.

`range(start, stop, step)` does **not** include the `stop` value.

````python
print("0 through 4:")
for i in range(5):
    print(i)

print("\n2 through 10 in steps of 2:")
for i in range(2, 11, 2):
    print(i)
````

## 17. `while` Loops

A `while` loop repeats as long as its condition remains `True`.

Always ensure that something changes inside the loop so that it eventually terminates.

````python
count = 1

while count <= 5:
    print(count)
    count += 1
````

## 18. `break`, `continue`, and `enumerate()`

- `break` exits a loop early.
- `continue` skips to the next iteration.
- `enumerate()` provides both the index and value while iterating.

These tools make loops easier to control.

````python
scores = [78, 85, -1, 92, 88]

for index, score in enumerate(scores):
    if score < 0:
        print(f"Invalid score at position {index}; skipping.")
        continue

    print(index, score)

    if score >= 90:
        print("Found a score of at least 90.")
        break
````

## 19. Functions

A **function** is a reusable block of code.

Functions improve organization, reduce repetition, and make programs easier to test. Machine Learning workflows frequently organize preprocessing, training, prediction, and evaluation into functions.

````python
def calculate_mean(values):
    total = sum(values)
    return total / len(values)

scores = [70, 80, 90, 100]
average = calculate_mean(scores)

print("Mean:", average)
````

## 20. Function Parameters and Default Arguments

Functions may accept multiple parameters. A parameter can also have a default value.

Use descriptive function and variable names so that code remains readable.

````python
def accuracy(correct, total, percentage=True):
    value = correct / total

    if percentage:
        return value * 100
    return value

print(accuracy(85, 100))
print(accuracy(85, 100, percentage=False))
````

## 21. Variable Scope

Variables created inside a function are generally **local** to that function.

Avoid excessive use of global variables. Passing values into functions and returning results usually produces clearer code.

````python
learning_rate = 0.01  # Global variable

def train_step(loss):
    updated_loss = loss * (1 - learning_rate)  # Local variable
    return updated_loss

print(train_step(2.5))
# print(updated_loss)  # This would produce a NameError outside the function.
````

## 22. List Comprehensions

A **list comprehension** creates a list concisely from another iterable.

They are useful for transformations and filtering, but very complicated comprehensions should be replaced with normal loops for readability.

````python
numbers = [1, 2, 3, 4, 5]

squares = [x ** 2 for x in numbers]
even_squares = [x ** 2 for x in numbers if x % 2 == 0]

print(squares)
print(even_squares)
````

Programs sometimes encounter invalid input or unexpected conditions.

## 27. Object-Oriented Programming: Classes and Objects

A **class** defines a type of object. An object can contain:

- **attributes** — stored data

You do not need advanced object-oriented programming to begin Machine Learning, but understanding classes helps when working with libraries such as scikit-learn, where models are represented as objects.

## 32. Vectorized Operations

NumPy performs operations on entire arrays without requiring explicit Python loops.

This is called **vectorization** and is fundamental to efficient numerical computing and Machine Learning.

````python
x = np.array([1, 2, 3, 4])

print("x:", x)
print("x + 10:", x + 10)
print("x * 2:", x * 2)
print("x squared:", x ** 2)
print("Square root:", np.sqrt(x))
````

## 33. Aggregation and Descriptive Statistics

NumPy provides convenient functions for summarizing numerical data, including:

## 34. Axis Operations

For matrices, the `axis` argument determines the direction of an aggregation.

- `axis=0` aggregates **down rows**, producing one result per column.
- `axis=1` aggregates **across columns**, producing one result per row.

````python
X = np.array([
    [1, 10],
    [2, 20],
    [3, 30]
])

print("Column means:", np.mean(X, axis=0))
print("Row means:", np.mean(X, axis=1))
````

## 37. Matrix Operations

Linear algebra is central to Machine Learning.

The `@` operator performs matrix multiplication, while `.T` computes the transpose.

````python
A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

print("A + B:")
print(A + B)

print("\nElement-wise multiplication:")
print(A * B)

print("\nMatrix multiplication:")
print(A @ B)

print("\nTranspose of A:")
print(A.T)
````

## 43. Missing Values

Real datasets frequently contain missing observations.

pandas represents many missing numeric values as `NaN`.

Common operations include:

- `isna()`
- `dropna()`
- `fillna()`

The correct treatment of missing data depends on the problem and should not be chosen automatically.

````python
missing_df = pd.DataFrame({
    "age": [20, 25, None, 40],
    "score": [80, None, 75, 92]
})

print("Missing-value counts:")
print(missing_df.isna().sum())

filled_df = missing_df.copy()
filled_df["age"] = filled_df["age"].fillna(filled_df["age"].mean())
filled_df["score"] = filled_df["score"].fillna(filled_df["score"].mean())

print("\nAfter mean imputation:")
display(filled_df)
````

## 52. Normalization and Standardization Intuition

Features can have very different numerical scales.

For example, age may range from 18–80 while income may range from thousands to hundreds of thousands.

A simple **standard score** transforms a value using:

\[
z = \frac{x-\mu}{\sigma}
\]

where:

- \(x\) is the original value
- \(\mu\) is the mean
- \(\sigma\) is the standard deviation

Many ML algorithms benefit from scaled features.

````python
x = np.array([10, 20, 30, 40, 50], dtype=float)

mean = x.mean()
std = x.std()

z = (x - mean) / std

print("Original:", x)
print("Mean:", mean)
print("Standard deviation:", std)
print("Standardized:", z)
print("Standardized mean:", z.mean())
````

## 53. Train/Test Splitting — Conceptual Example

A Machine Learning model should be evaluated on data it did not use for training.

A common workflow divides observations into:

- **training set** — used to fit the model
- **test set** — used to evaluate generalization

Here we demonstrate the idea using NumPy before introducing specialized ML libraries.

````python
rng = np.random.default_rng(42)

X = np.arange(20).reshape(10, 2)
y = np.arange(10)

indices = rng.permutation(len(X))
train_size = 8

train_indices = indices[:train_size]
test_indices = indices[train_size:]

X_train = X[train_indices]
X_test = X[test_indices]
y_train = y[train_indices]
y_test = y[test_indices]

print("Training X shape:", X_train.shape)
print("Test X shape:", X_test.shape)
print("Training indices:", train_indices)
print("Test indices:", test_indices)
````

## 54. A Simple Prediction Rule and Accuracy

Before using a formal ML model, we can understand prediction and evaluation with a simple rule.

**Accuracy** is:

accuracy = number of correct predictions / total predictions

Accuracy is useful for some classification problems, although it can be misleading for highly imbalanced datasets.

````python
ages = np.array([18, 25, 31, 44, 52, 60])
actual = np.array([0, 0, 0, 1, 1, 1])

# Simple hand-written prediction rule:
predicted = (ages >= 40).astype(int)

accuracy_value = np.mean(predicted == actual)

print("Actual:   ", actual)
print("Predicted:", predicted)
print(f"Accuracy: {accuracy_value:.2%}")
````

## 55. Optional: First scikit-learn Model

If `scikit-learn` is installed, the following example trains a very small linear regression model.

The important API pattern is:

1. Create a model object.
2. Call `.fit(X, y)` to train it.
